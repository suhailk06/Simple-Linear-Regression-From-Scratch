```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
dataset=pd.read_csv("Data.csv")
X=dataset.iloc[:,1:2].values
Y=dataset.iloc[:,2:3].values
```

```python
print(X)
```

```python
print(Y)
```

```python
#slope(m)=((n∑(XY)-(∑X)(∑Y)))/((n∑(X^2 )-(∑X)^2 ))
xy=0
x=0
y=0
x2=0
y2=0
n=len(X)
for i in range(len(X)):
  x+=X[i]
  y+=Y[i]
  x2+=X[i]**2
  y2+=Y[i]**2
  xy+=X[i]*Y[i]


print("Calculate the Slope (m):")
print(f"n={n}")
print(f"∑xy={xy}")
print(f"n*(∑xy)={n*(xy)}")
print(f"∑x={x}")
print(f"∑y={y}")
print(f"∑x2={x2}")
print(f"n*(∑x2)={n*x2}")
print(f"(∑x)**2={x**2}")
m=((n*(xy))-(x*y))/((n*x2)-(x**2))
print(f"slope(m)={m}")
```

```python
#Intercept(c)=(∑Y-m(∑X))/n

print("Calculate the Intercept (c):")
print(f"∑y={y}")
print(f"m(∑x)={m*x}")
print(f"n={n}")
c=(y-(m*x))/n
print(f"Intercept(c)={c}")
```

```python
print("Regression Equation")
print("Y=m*X+c")
print(f"slope(m)={m}")
print(f"Intercept(c)={c}")
print(f"Regression Equation Y={m}*x+{c}")
```

```python
#Correlation Coefficient(r)=(n∑(XY)-(∑X)(∑Y))/√([n∑(X^2 )-(∑X)^2 ]×[n∑(Y^2 )-(∑Y)^2 ] )
print("Calculate the Correlation Coefficient (r):")

print(f"n*(xy)={n*(xy)}")
print(f"(∑x)(∑y)={x*y}")
print(f"n*(x2)={n*x2}")
print(f"(∑x)**2={x**2}")
print(f"n*(y2)={n*y2}")
print(f"(∑y)**2={y**2}")

r=((n*(xy))-(x*y))/((((n*x2)-(x**2))*((n*y2)-(y**2)))**0.5)
print(f"Correlation Coefficient(r)={r}")
```
